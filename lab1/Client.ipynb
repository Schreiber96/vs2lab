# Running server and client in separate notebooks - client part

Here we start the client. The server needs to be started first. Lets go....

````python
import myClientServer as clientserver # import from our script

#client = clientserver.Client() # Create a client instance
def client_get(name):
    client = clientserver.Client()
    client.get(name)
    client.close

def client_add(name, number):
    client = clientserver.Client()
    client.addToDictionary(name, number)
    client.close

client_add("Bob", "0187")
client_add("Rob", "0123")
client_add("Bobs", "0187T")
client_add("Bo:b", "018723")
client_get("ali")
client_get("Bob")
client_get(" ")
client_get("  Rob  ")
client = clientserver.Client()
client.getAll()

#client.call() # start the call and print echo
````

## Telefonbuch Protokoll

Das Protokoll beginnt mit der Art der Nachricht, anschließend folgt als Trenner ein Semikolon. Bei notwendigen Paramteren werden diese nachfolgend angehängt und ebenfalls mit einem Semikolon betrennt. Ein Semikolon schließt die Nachricht ab.

Besteht ein Paramteter aus einer Liste werden die einzelnen Elemente mit einem Komma getrennt.

Für das Protokoll sind folgende Funktionen definiert:
```
Add;Name;
Get;Name;
Getall;
Return;Nummer;
ReturnAll;Name,Nummer;Name,Nummer;...;
Error;ErrorMsg;
```

### Vorgehen
Bevor ein Befehl ausgeführt werden muss, muss zuerst ein Client erstellt werden.

````python
import myClientServer as clientserver # import from our script
client = clientserver.Client()
````

Danach kann dann auf dem client-Objekt die gewünschte Methode (get, getall, addToDictionary) ausgeführt werden.

````python
client.get("name")
# oder
# client.getall()
# oder
# client.addToDictionary("name", "number")
````

Anschließend muss das client-Objekt wieder geschlossen werden.

````python
client.close()
````

Aus Komfort-gründen stehen die Hilfsfunktionen client_get() und client_add() zur Verfügung, die sich um die Erstellung und Schließung des Klienten kümmern.
